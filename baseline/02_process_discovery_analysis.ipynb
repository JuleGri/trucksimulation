VARIANT ANALYSIS

```python
from pm4py.statistics.traces.generic.log import case_statistics

variants = case_statistics.get_variant_statistics(
    log
)

variants = sorted(
    variants,
    key=lambda x: x["count"],
    reverse=True
)

for v in variants[:10]:
    print(v)

print(
    "Number of variants:",
    len(variants)
)
```

```python
total_cases = sum(
    v["count"]
    for v in variants
)

top3_cases = sum(
    v["count"]
    for v in variants[:3]
)

print(
    "Top-3 coverage:",
    round(
        top3_cases / total_cases * 100,
        2
    ),
    "%"
)
```

```python
dfg, start, end = pm4py.discover_dfg(
    log
)

print(
    sorted(
        dfg.items(),
        key=lambda x: x[1],
        reverse=True
    )[:20]
)

# --- Persist discovery artifacts into discovery_params/<paramset>/archiv ---
import os
import pandas as pd
from collections import Counter
from datetime import datetime

# choose a parameter-set folder: prefer existing depth10_params_* if present
base_dir = os.path.join(os.getcwd(), '')  # notebook cwd is scenarios/baseline
param_root = os.path.join('discovery_params')
os.makedirs(param_root, exist_ok=True)
candidates = [p for p in os.listdir(param_root) if os.path.isdir(os.path.join(param_root, p))]
paramset = None
for c in candidates:
    if 'depth10' in c or 'params' in c:
        paramset = c
        break
if paramset is None:
    paramset = 'params_' + datetime.now().strftime('%Y%m%d_%H%M%S')
    os.makedirs(os.path.join(param_root, paramset), exist_ok=True)
archiv_dir = os.path.join(param_root, paramset, 'archiv')
os.makedirs(archiv_dir, exist_ok=True)

# Save top DFG edges to CSV
dfg_items = sorted(dfg.items(), key=lambda x: x[1], reverse=True)
dfg_df = pd.DataFrame([{'edge_from':e[0][0], 'edge_to':e[0][1], 'count':e[1]} for e in dfg_items])
dfg_df.to_csv(os.path.join(archiv_dir, 'dfg_top_edges.csv'), index=False)

# Save variants and coverage
variants_df = pd.DataFrame([{'variant': tuple(v['variant']), 'count': v['count']} for v in variants])
variants_df = variants_df.sort_values('count', ascending=False).reset_index(drop=True)
variants_df['cum_coverage'] = variants_df['count'].cumsum() / variants_df['count'].sum()
variants_df.to_csv(os.path.join(archiv_dir, 'variants_top.csv'), index=False)

# Routing probabilities (next activity given current activity)
next_transitions = []
for trace in log:
    acts = [e['concept:name'] for e in trace if 'concept:name' in e]
    for i in range(len(acts)-1):
        next_transitions.append((acts[i], acts[i+1]))
c = Counter(next_transitions)
rows = []
totals = {}
for (a,b),count in c.items():
    totals.setdefault(a,0)
    totals[a]+=count
for (a,b),count in c.items():
    rows.append({'from':a,'to':b,'count':count,'probability':count/totals[a]})
pd.DataFrame(rows).to_csv(os.path.join(archiv_dir, 'discovered_routing_probabilities.csv'), index=False)

# Export discovered Petri net visualization if pm4py visualization available
try:
    from pm4py.visualization.petrinet import visualizer as pn_visualizer
    gviz = pn_visualizer.apply(net, im, fm)
    pn_visualizer.save(gviz, os.path.join(archiv_dir, 'discovered_petri_net.png'))
    print('Saved discovered_petri_net.png to', archiv_dir)
except Exception as e:
    print('Could not export petri net visualization:', e)

# Attempt to load existing empirical activity durations CSV and clean waiting times if present
emp_path = os.path.join(os.getcwd(), 'empirical_activity_durations.csv')
if os.path.exists(emp_path):
    try:
        # best-effort read: skip the multi-header lines if present
        emp = pd.read_csv(emp_path, header=2, index_col=0)
    except Exception:
        emp = pd.read_csv(emp_path, index_col=0)
    # normalize column names if duplicated
    emp = emp.loc[~emp.index.isna()]
    # Clip waiting time columns to non-negative where names contain 'waiting'
    waiting_cols = [c for c in emp.columns if 'waiting' in c.lower()]
    for wc in waiting_cols:
        emp[wc] = pd.to_numeric(emp[wc], errors='coerce').clip(lower=0)
    # Drop extreme outliers at 99.9 percentile per waiting column
    for wc in waiting_cols:
        cutoff = emp[wc].quantile(0.999)
        emp = emp[emp[wc] <= cutoff]
    emp.to_csv(os.path.join(archiv_dir, 'empirical_activity_durations_cleaned.csv'))
    print('Saved cleaned empirical activity durations to archiv')
else:
    print('No empirical_activity_durations.csv found in current folder; skipping cleaning step')

# Estimate resource concurrency if resource attribute exists
try:
    # collect intervals if lifecycle:start and lifecycle:complete are present, else use event timestamp as instantaneous
    intervals = {}  # pool -> list of (start, end)
    for trace in log:
        for ev in trace:
            res = ev.get('org:resource') or ev.get('resource') or ev.get('org_resource')
            if res is None:
                continue
            ts = ev.get('time:timestamp') or ev.get('timestamp')
            if ts is None:
                continue
            # treat as instantaneous point for concurrency counting
            intervals.setdefault(res, []).append(pd.to_datetime(ts))
    # compute simple concurrency: max number of events at the same second per resource
    resource_concurrency = []
    for r, times in intervals.items():
        if len(times)==0:
            continue
        times = pd.to_datetime(times)
           # DatetimeIndex doesn't have .dt; use floor directly
   floored = times.floor('s') if hasattr(times, 'floor') else pd.to_datetime(times).to_series().dt.floor('s')
   counts = pd.Series(floored).value_counts()
        resource_concurrency.append({'resource':r, 'estimated_max_concurrency': int(counts.max())})
    pd.DataFrame(resource_concurrency).to_csv(os.path.join(archiv_dir, 'resource_estimated_concurrency.csv'), index=False)
    print('Saved resource_estimated_concurrency.csv')
except Exception as e:
    print('Resource concurrency estimation skipped or failed:', e)

print('Discovery artifacts saved under', archiv_dir)
```
