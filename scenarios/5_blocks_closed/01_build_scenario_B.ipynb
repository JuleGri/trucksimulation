```python
import pickle 
import copy 

with open(
    "../baseline/baseline.pkl",
    "rb"
) as f:
    params = pickle.load(f)

print(params.act_to_resources)
```

```python
scenario_params = copy.deepcopy(params)

closed_blocks = [
    "T22",
    "T23",
    "T24",
    "T25",
    "T26"
]
```

```python
#DEACTIVATE RESOURCES
scenario_params.resources = [

    r

    for r in scenario_params.resources

    if r not in closed_blocks

]
print(sorted(scenario_params.resources))
```

```python
#REMOVE CONCURRENCY
for block in closed_blocks:

    scenario_params.max_concurrency.pop(
        block,
        None
    )

    scenario_params.resource_weights.pop(
        block,
        None
    )
    print(
scenario_params.max_concurrency.keys()
)
```

```python
#ACTIVITY RESOURCE MAPPING
for act,res in (
    scenario_params.act_to_resources.items()
):

    scenario_params.act_to_resources[act] = [

        r

        for r in res

        if r not in closed_blocks

    ]
```

```python
# Check
print(
    scenario_params.act_to_resources[
        "RMG_receive"
    ]
)

print(
    scenario_params.act_to_resources[
        "RMG_delivery"
    ]
)

print(
    scenario_params.act_to_resources[
        "RMG_mixed"
    ]
)
```

```python
#Safety Check - no output expected
for act, res_list in (
    scenario_params.act_to_resources.items()
):

    if len(res_list) == 0:

        print(
            "EMPTY:",
            act
        )
```

```python
# document ressource status
print(
    "Baseline resources:",
    len(params.resources)
)

print(
    "Scenario resources:",
    len(scenario_params.resources)
)
```

```python
# build test simulation
from prosit.simulator import SimulatorEngine

scenario_engine = SimulatorEngine(
    scenario_params
)

scenario_log = scenario_engine.apply(
    n_traces=20000
)
```

```python
scenario_log.to_csv(
    "scenario_B_5_blocks_closed_sim_log.csv",
    index=False
)
```
