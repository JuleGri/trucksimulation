```python
import copy

scenario_params = copy.deepcopy(params)

closed_blocks = [
    "T22",
    "T23",
    "T24"
]

scenario_params.resources = [

    r

    for r in scenario_params.resources

    if r not in closed_blocks

]

print("Baseline:")
print(len(params.resources))
print(params.resources)

print("\nScenario:")
print(len(scenario_params.resources))
print(scenario_params.resources)
```
