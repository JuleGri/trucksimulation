```python
import pickle 

with open(
    "../../data/processed/CTB/prosit_params/sim_log_s6_sample_30.000_depth10_params_HO2_LL_2.pkl",
    "rb"
) as f:

    scenario_params = pickle.load(f)

for act in sorted(
    scenario_params.act_to_resources.keys()
):
    print(act)
print(scenario_params.act_to_resources)
```

```python
for act in [
    "RMG_receive",
    "RMG_delivery",
    "RMG_mixed"
]:

    print("\n", act)

    print(
        scenario_params.act_to_resources[act]
    )
```

```python
# ==========================================================
# SCENARIO B
# T20 + T21 + T22 CLOSED
# ==========================================================

closed_blocks = [
    "T20",
    "T21",
    "T22"
]

for activity in [
    "RMG_receive",
    "RMG_delivery",
    "RMG_mixed"
]:

    scenario_params.act_to_resources[activity] = [

        r
        for r in scenario_params.act_to_resources[activity]

        if r not in closed_blocks
    ]

# ----------------------------------------------------------
# REMOVE RESOURCE SETTINGS
# ----------------------------------------------------------

for block in closed_blocks:

    scenario_params.max_concurrency.pop(
        block,
        None
    )

    scenario_params.calendars.pop(
        block,
        None
    )

print("Closed blocks:")
print(closed_blocks)

print("\nRemaining RMG resources:")

for activity in [
    "RMG_receive",
    "RMG_delivery",
    "RMG_mixed"
]:
    print(
        f"\n{activity}:"
    )
    print(
        scenario_params.act_to_resources[
            activity
        ]
    )
```
