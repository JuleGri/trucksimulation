```python
import pandas as pd

baseline_sim = pd.read_csv("../../data/processed/CTB/prosit_simulations/sim_log_s6_sample_24.000_depth10_maxConc_2.csv")
scenario_sim = pd.read_csv("../../data/processed/CTB/prosit_simulations/scenario_a/sim_log_24.000_T20_closed.csv")
```

```python
# Konttrolle
baseline_res = (
    baseline_sim["org:resource"]
    .value_counts()
    .rename("baseline")
)

scenario_res = (
    scenario_sim["org:resource"]
    .value_counts()
    .rename("scenario")
)

relocation = pd.concat(
    [
        baseline_res,
        scenario_res
    ],
    axis=1
).fillna(0)

relocation["diff"] = (
    relocation["scenario"]
    - relocation["baseline"]
)

relocation.sort_values(
    "diff",
    ascending=False
)
```
