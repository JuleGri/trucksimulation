```python
import pickle 

with open(
    "../../data/processed/CTB/prosit_params/sim_log_s6_sample_30.000_depth10_params_HO2_LL_2.pkl",
    "rb"
) as f:

    scenario_params = pickle.load(f)

for act in sorted(
    scenario_params.act_to_resources.keys()
):
    print(act)
print(scenario_params.act_to_resources)
```

```python
for act in [
    "RMG_receive",
    "RMG_delivery",
    "RMG_mixed"
]:

    print("\n", act)

    print(
        scenario_params.act_to_resources[act]
    )
```

```python
#Remove resource T20
for activity in [
    "RMG_receive",
    "RMG_delivery",
    "RMG_mixed"
]:

    scenario_params.act_to_resources[activity] = [

        r for r in
        scenario_params.act_to_resources[activity]

        if r != "T20"
    ]
scenario_params.max_concurrency.pop(
    "T20",
    None
)
scenario_params.calendars.pop(
    "T20",
    None
)
```

```python
import pickle

with open(
    "../../data/processed/CTB/prosit_simulations/scenario_a/scenario_A_T20_closed.pkl",
    "wb"
) as f:

    pickle.dump(
        scenario_params,
        f
    )
```

```python
# build test simulation
from prosit.simulator import SimulatorEngine

scenario_engine = SimulatorEngine(
    scenario_params
)

scenario_log = scenario_engine.apply(
    n_traces=24000
)
```
