```python
import pandas as pd

ref = pd.read_csv("../../data/processed/CTB/s7_eventlog_with_performance_targets.csv")
```

```python
for col in [
    "enabled:timestamp",
    "start:timestamp",
    "time:timestamp"
]:
    ref[col] = pd.to_datetime(
        ref[col]
    )
```

```python
arrival_times = (
    ref.groupby(
        "case:concept:name"
    )["start:timestamp"]
    .min()
)

hourly_arrivals = (
    arrival_times.dt.hour
    .value_counts()
    .sort_index()
)

hourly_arrivals
```

```python
weekday_arrivals = (
    arrival_times.dt.dayofweek
    .value_counts()
    .sort_index()
)

weekday_arrivals
```
